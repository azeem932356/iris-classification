```python
from sklearn.datasets import load_iris

iris= load_iris(as_frame=True)
dir(iris)
```

```python
#feature names

iris.feature_names
```

```python
iris.target_names
```

```python
iris.data
```

```python
#converting into dataframe

df= iris.frame.copy()
df.head(5)
```

```python
#our unique values, target has those

df.target.unique()
```

#### Data Cleaning

```python
df.columns= ['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'target']
df.head(5)
```

```python
#instead of this target number system we are replacing with thier names

#first look what are the names

iris.target_names
```

```python
#converting and adding new column based on conditions 

df['species']= df['target'].map(dict(enumerate(iris.target_names)))
df.head(5)
```

```python
#droping target colum

df= df.drop(columns= ['target'])
df.head()
```

```python
df.species.value_counts()
```

#### Ploting data using matplotlib

```python
import matplotlib.pyplot as plt
```

```python
colors= {"setosa":"#5BB8EB", "versicolor":"#39D353", "virginica":"#e6d9f4"}

for name, group in df.groupby('species'):
    plt.scatter(group.petal_length, group.petal_width, color= colors[name], edgecolor= 'k', label= name)

plt.xlabel("petal length (cm)")
plt.ylabel("petal width (cm)")
plt.title("petal seperate the species pretty well")
plt.legend()
plt.show()
```

```python
#same with sepal

colors= {"setosa":"#5BB8EB", "versicolor":"#39D353", "virginica":"#e6d9f4"}

for name, group in df.groupby('species'):
    plt.scatter(group.sepal_length, group.sepal_width, color= colors[name], edgecolor= 'k', label= name)

plt.xlabel("sepal length (cm)")
plt.ylabel("sepal width (cm)")
plt.title("sepal")
plt.legend()
plt.show()
```

```python
df.head()
```

```python
#initilizing the X and y

features= ['sepal_length', 'sepal_width', 'petal_length', 'petal_width']

X= df[features]
y= df.species
```

```python
#train_test_split

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=11
)

print("Training rows:", X_train.shape[0])
print("Testing rows:", X_test.shape[0])
```

```python
#model training

from sklearn.linear_model import LogisticRegression

model= LogisticRegression()
model.fit(X_train, y_train)
```

```python
#to see what model is learned like things

model.classes_
```

```python
X_train.shape[0]
```

```python
import pandas as pd
```

```python
#testing on test data and comparing with predicted values

y_pred= model.predict(X_test)

compare= pd.DataFrame({'actual': y_test, 'predicted': y_pred})
compare.head(10)
```

```python
#confusion matrix is used to determine in numbers how many actually it got wrong and how many it got right in a square 

from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay

cm= confusion_matrix(y_test, y_pred, labels= model.classes_)
print(cm)
```

```python
#visual of confusionmatrix

import matplotlib.pyplot as plt

disp= ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=model.classes_)
disp.plot(cmap= 'Blues', colorbar= False)
plt.title("Confusion Matrix (test_set)")
plt.show()
```

```python
#report

from sklearn.metrics import classification_report

print(classification_report(y_test, y_pred))
```
